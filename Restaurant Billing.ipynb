```python
customer_name = input("Enter customer name: ")

print("1 - In-Dining")
print("2 - Takeaway")

order_type = int(input("Enter choice: "))

if order_type == 1:
    table = input("Enter your table no: ")
    packing = 0
else:
    table = "N/A"
    packing = 50


items = []
prices = []
quantity = []

choice = 1


while choice != 0:

    print("\n1. Food")
    print("2. Beverages and Desserts")
    print("0. BILL")

    choice = int(input("Enter choice: "))

    if choice == 1:

        print("1. Paneer Tikka - Rs 260")
        print("2. Butter Chicken - Rs 320")
        print("3. Dal Makhani - Rs 220")
        print("4. Butter Naan - Rs 80")

        item = int(input("Enter item: "))
        qty = int(input("Enter Quantity: "))

        if item == 1:
            items.append("Paneer Tikka")
            prices.append(260)
            quantity.append(qty)

        elif item == 2:
            items.append("Butter Chicken")
            prices.append(320)
            quantity.append(qty)

        elif item == 3:
            items.append("Dal Makhani")
            prices.append(220)
            quantity.append(qty)

        elif item == 4:
            items.append("Butter Naan")
            prices.append(80)
            quantity.append(qty)


    elif choice == 2:

        print("1. Cold Coffee - Rs 80")
        print("2. Masala Chai - Rs 40")
        print("3. Gulab Jamun - Rs 100")

        item = int(input("Enter item: "))
        qty = int(input("Enter Quantity: "))

        if item == 1:
            items.append("Cold Coffee")
            prices.append(80)
            quantity.append(qty)

        elif item == 2:
            items.append("Masala Chai")
            prices.append(40)
            quantity.append(qty)

        elif item == 3:
            items.append("Gulab Jamun")
            prices.append(100)
            quantity.append(qty)


    elif choice == 0:
        print("The final bill is:")


subtotal = 0

print("\nFinal Bill")
print("Customer:", customer_name)
print("Table:", table)

for i in range(len(items)):
    amount = prices[i] * quantity[i]
    subtotal = subtotal + amount
    print(items[i], quantity[i], amount)


gst = subtotal * 18 / 100

total = subtotal + gst + packing


print("Subtotal: Rs.", subtotal)
print("GST: Rs.", gst)
print("Packing: Rs.", packing)
print("Total: Rs.", total)
```
